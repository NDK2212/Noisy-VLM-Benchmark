```python
import os
import json
import warnings
import ast
import pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed
from kaggle_secrets import UserSecretsClient
from openai import AzureOpenAI
from tqdm import tqdm

warnings.filterwarnings("ignore")

# ==========================================
# 1. COCO METRICS IMPORTS
# ==========================================
from pycocoevalcap.tokenizer.ptbtokenizer import PTBTokenizer
from pycocoevalcap.bleu.bleu import Bleu
from pycocoevalcap.rouge.rouge import Rouge
from pycocoevalcap.cider.cider import Cider
from pycocoevalcap.meteor.meteor import Meteor
# Spice is completely removed for speed!

# ==========================================
# 2. BULLETPROOF AZURE CONNECTION
# ==========================================
def build_azure_client(api_key: str, azure_endpoint: str, api_version: str, timeout_seconds: int = 120, max_retries: int = 5):
    """Constructs the Azure client."""
    return AzureOpenAI(
        api_key=api_key,
        azure_endpoint=azure_endpoint,
        api_version=api_version,
        timeout=timeout_seconds,
        max_retries=max_retries
    )

# ==========================================
# 3. CLAIR SCORER LOGIC
# ==========================================
def paper_accurate_clair_azure(candidate_caption, reference_captions, client, deployment_name):
    prompt_template = """\
You are trying to tell if a candidate set of captions is describing the same image as a reference set of captions.
Candidate set:
{candidate_statements}
Reference set:
{target_statements}
On a precise scale from 0 to 100, how likely is it that the candidate set is \
describing the same image as the reference set? (JSON format, with a key "score", \
value between 0 and 100, and a key "reason" with a string value.)"""

    formatted_prompt = prompt_template.format(
        candidate_statements=str([candidate_caption]),
        target_statements=str(reference_captions)
    )

    try:
        response = client.chat.completions.create(
            model=deployment_name,
            response_format={ "type": "json_object" },
            messages=[
                {"role": "system", "content": "You are a precise evaluation system. You must output valid JSON only."},
                {"role": "user", "content": formatted_prompt}
            ],
            temperature=0.0
        )
        
        result = json.loads(response.choices[0].message.content)
        score = float(result.get("score", 0.0))
        return score / 100.0 if score > 1.0 else score
        
    except Exception as e:
        # Silenced the error print here so it doesn't spam during multi-threading
        return 0.0

# ==========================================
# 4. FULL PIPELINE EXECUTION
# ==========================================
def evaluate_all_metrics(predictions, references, client, deployment_name):
    gts = {img_id: [{'caption': cap} for cap in caps] for img_id, caps in references.items()}
    res = {img_id: [{'caption': caps[0]}] for img_id, caps in predictions.items()}
    
    tokenizer = PTBTokenizer()
    gts = tokenizer.tokenize(gts)
    res = tokenizer.tokenize(res)
    
    scorers = [
        (Bleu(4), ["Bleu_1", "Bleu_2", "Bleu_3", "Bleu_4"]),
        (Meteor(), "METEOR"),
        (Rouge(), "ROUGE_L"),
        (Cider(), "CIDEr")
    ]
    
    final_scores = {}
    
    # Calculate Traditional Metrics (Removed tqdm here to keep console clean for multiple files)
    for scorer, method in scorers:        
        try:
            avg_score, _ = scorer.compute_score(gts, res)
            
            if isinstance(method, list): 
                for idx, sc in enumerate(avg_score):
                    final_scores[method[idx]] = sc
            else:
                final_scores[method] = avg_score
                            
        except Exception as e:
            print(f"❌ Error computing {method}: {e}")

    # Calculate CLAIR
    if not client or not deployment_name:
        final_scores["CLAIR"] = None
    else:
        clair_scores = []
        total_items = len(predictions)
        
        def evaluate_single_clair(img_id):
            candidate = predictions[img_id][0]
            refs = references[img_id]
            return paper_accurate_clair_azure(candidate, refs, client, deployment_name)

        with ThreadPoolExecutor(max_workers=10) as executor:
            future_to_score = {
                executor.submit(evaluate_single_clair, img_id): img_id 
                for img_id in predictions.keys()
            }
            
            for future in tqdm(as_completed(future_to_score), total=total_items, desc="  CLAIR Eval", leave=False):
                try:
                    score = future.result()
                    clair_scores.append(score)
                except Exception as e:
                    clair_scores.append(0.0)
            
        final_scores["CLAIR"] = sum(clair_scores) / len(clair_scores) if clair_scores else 0.0

    return final_scores

# ==========================================
# 5. MAIN SCRIPT
# ==========================================
if __name__ == "__main__":
    print("="*60)
    print(" 📂 BATCH FOLDER EVALUATION BOOT SEQUENCE 📂 ")
    print("="*60)

    # 1. Pull Azure Secret
    try:
        user_secrets = UserSecretsClient()
        api_key = user_secrets.get_secret("AZURE_API_KEY")
        endpoint = user_secrets.get_secret("AZURE_ENDPOINT")
        print("✅ Secret 'AZURE_API_KEY' loaded successfully.")
        print("✅ Secret 'AZURE_ENDPOINT' loaded successfully.")
    except Exception as e:
        print(f"❌ Could not load Kaggle Secret 'AZURE_API_KEY': {e}")
        api_key = ""

    RUN_CONFIG = {
        "deployment_name": "gpt-4o",
        "api_version": "2024-12-01-preview",
        "timeout_seconds": 120,
        "max_retries": 5
    }

    client = None
    if api_key and endpoint:
        client = build_azure_client(
            api_key=api_key, azure_endpoint=endpoint,
            api_version=RUN_CONFIG["api_version"], timeout_seconds=RUN_CONFIG["timeout_seconds"],
            max_retries=RUN_CONFIG["max_retries"]
        )
    else:
        print("❌ Cannot build Azure client without API key. CLAIR will be skipped.")

    # ==========================================
    # FOLDER PROCESSING LOGIC
    # ==========================================
    
    # 1. Define your 9 datasets and their corresponding subfolders based on your image
    datasets_to_process = [
        ("dam-paco-image-bbox", "image_bbox"),
        ("dam-paco-image-mask", "image_mask"),
        ("dam-paco-mask-noise", "mask"),
        ("gpt4roi-paco-bbox-noise", "bbox_noise"),
        ("gpt4roi-paco-image-bbox", "image_noise"),
        ("shikra-paco-bbox-noise", "bbox"),
        ("shikra-paco-image-bbox", "image"),
        ("vip-llava-paco-bbox-noise", "bbox"),
        ("vip-llava-paco-image-bbox", "image")
    ]

    BASE_INPUT_DIR = "/kaggle/input/datasets/ckhangnguyn"

    # 2. Master Loop: Go through each of the 9 datasets
    for dataset_slug, folder_name in datasets_to_process:
        print("\n" + "★"*60)
        print(f"🚀 STARTING DATASET: {dataset_slug.upper()}")
        print("★"*60)

        INPUT_PATH = f"{BASE_INPUT_DIR}/{dataset_slug}/{folder_name}"
        OUTPUT_CSV_NAME = f"evaluation_summary_{dataset_slug}.csv"
        
        # IMPORTANT: Reset the results list for each dataset so they don't mix!
        all_results_data = []

        # Smart Detection: Is it a single file or a folder?
        if os.path.exists(INPUT_PATH):
            if os.path.isfile(INPUT_PATH) and INPUT_PATH.endswith(".csv"):
                INPUT_FOLDER = os.path.dirname(INPUT_PATH)
                csv_files = [os.path.basename(INPUT_PATH)]
                print(f"\n📁 Found 1 specific CSV file: {csv_files[0]}")
            else:
                INPUT_FOLDER = INPUT_PATH
                csv_files = [f for f in os.listdir(INPUT_FOLDER) if f.endswith(".csv")]
                print(f"\n📁 Found {len(csv_files)} CSV files in {INPUT_FOLDER}")
        else:
            print(f"⚠️ Path not found, skipping: {INPUT_PATH}")
            continue

        for file_name in csv_files:
            file_path = os.path.join(INPUT_FOLDER, file_name)
            print(f"\n⏳ Processing: {file_name} ...")
            
            model_predictions = {}
            ground_truth_labels = {}
            file_metadata = {}

            try:
                df = pd.read_csv(file_path)

                meta_columns = [
                    "image_noise_type", "image_severity", 
                    "bbox_noise_type", "bbox_severity", 
                    "mask_noise_type", "mask_severity"
                ]
                for col in meta_columns:
                    if col in df.columns:
                        file_metadata[col] = df[col].iloc[0]
                    else:
                        file_metadata[col] = "N/A"
                
                PRED_COL = "pred_detailed"
                REF_COL = "caption"        
                for idx, row in df.iterrows():
                    unique_row_id = f"row_{idx}"
                    pred_str = str(row[PRED_COL])
                    model_predictions[unique_row_id] = [pred_str]
                    
                    refs = []
                    ref_str = str(row[REF_COL])
                    
                    if ref_str.strip().startswith('[') and ref_str.strip().endswith(']'):
                        try:
                            parsed_refs = ast.literal_eval(ref_str)
                            if isinstance(parsed_refs, list):
                                refs.extend([str(r) for r in parsed_refs])
                            else:
                                refs.append(ref_str)
                        except (ValueError, SyntaxError):
                            refs.append(ref_str)
                    else:
                        refs.append(ref_str)
                            
                    ground_truth_labels[unique_row_id] = refs
                    
                # Execute the pipeline for this specific file
                results = evaluate_all_metrics(
                    predictions=model_predictions, 
                    references=ground_truth_labels,
                    client=client,
                    deployment_name=RUN_CONFIG["deployment_name"]
                )
                
                # Attach the filename to the results dictionary
                results["Filename"] = file_name
                results.update(file_metadata)
                all_results_data.append(results)
                print(f"✅ Finished {file_name}")

            except Exception as e:
                print(f"❌ Error processing {file_name}: {e}")

        # ==========================================
        # 6. EXPORT TO CSV (Happens 9 times, once per dataset)
        # ==========================================
        if all_results_data:
            print("\n💾 EXPORTING FINAL RESULTS FOR: " + dataset_slug)
            
            df_out = pd.DataFrame(all_results_data)
            
            # Define the exact order for the front columns
            front_cols = [
                "Filename", "image_noise_type", "image_severity", 
                "bbox_noise_type", "bbox_severity", "mask_noise_type", "mask_severity"
            ]
            
            # Keep only the columns that actually made it into the dataframe
            front_cols = [c for c in front_cols if c in df_out.columns]
            
            # Grab all the metric columns (Bleu, METEOR, CLAIR, etc.)
            metric_cols = [c for c in df_out.columns if c not in front_cols]
            
            # Reorder the dataframe
            df_out = df_out[front_cols + metric_cols]
            
            # Save to the current working directory
            df_out.to_csv(OUTPUT_CSV_NAME, index=False)
            print(f"🎉 Successfully saved: {OUTPUT_CSV_NAME}")
        else:
            print(f"\n⚠️ No valid CSV files were processed for {dataset_slug}. Output table was not generated.")

    print("\n" + "="*60)
    print(" 🏁 ALL DATASETS PROCESSED SUCCESSFULLY! 🏁 ")
    print("="*60)
```
